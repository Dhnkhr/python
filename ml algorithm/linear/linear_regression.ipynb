```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
%matplotlib inline
```

```python
df=pd.read_csv('dummy_height_weight_large.csv')
df.head()
```

```python
plt.scatter(df['weight_kg'],df['height_cm'])
```

```python
##corelation
```

```python
df.corr()
```

```python
#independent and independent features
df.head()
```

```python
X=df[['weight_kg']] ##independent features should be in dataframe or should be in the 2d array
X
```

```python
type(X)
```

```python
np.array(X).shape
```

```python
Y=df['height_cm'] ##dependent can be in seriers form or 1d array
Y
```

```python
## train test split
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test=train_test_split(X,Y,test_size=0.25,random_state=42)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
##standardisation
from sklearn.preprocessing import StandardScaler
scaler =StandardScaler()
X_train=scaler.fit_transform(X_train)
X_test=scaler.transform(X_test)
```

```python
## apply simple linear regression
from sklearn.linear_model import LinearRegression
regression = LinearRegression()
regression.fit(X_train,Y_train)
```

```python
regression
```

```python
print("slope: ",regression.coef_)
print("intercept: ", regression.intercept_)
```

```python
plt.scatter(X_train,Y_train)
plt.plot(X_train,regression.predict(X_train))
```

```python
##prediction for test data
y_predict=regression.predict(X_test)
```

```python
y_predict
```

```python
##performance
from sklearn.metrics import mean_absolute_error,mean_squared_error
```

```python
mse=mean_squared_error(Y_test,y_predict)
mae=mean_absolute_error(Y_test,y_predict)
rmse=np.sqrt(mse)
print(mse)
print(mae)
print(rmse)
```

```python
from sklearn.metrics import r2_score
```

```python
score=r2_score(Y_test,y_predict)
print(score)
```

```python
import statsmodels.api as sm
model=sm.OLS(Y_train,X_train).fit()
pred=model.predict(X_test)
print(pred) 
```
