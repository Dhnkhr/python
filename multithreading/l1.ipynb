```python
import time
import threading

def print_num() : 
    for i in range(5):
        time.sleep(2)
        print(i)

def print_char():
    for i in "abcde":
        time.sleep(2)
        print(i)

t1=threading.Thread(target=print_char)
t2=threading.Thread(target=print_num)

t=time.time()

t1.start()
t2.start()

t1.join()
t2.join()

t3=time.time()-t

print(t3)
```
