```python
import sqlite3
```

```python
connection=sqlite3.connect('example.db')
connection
```

```python
cursor=connection.cursor()
```

```python
cursor.execute('''
Create Table If Not Exists employee(
id integer Primary Key,
name Text Not Null,
age Integer,
department Text
) 
''')

connection.commit()
```

```python
cursor.execute('''
select * from employee
''')
```

```python
cursor.execute('''
Insert Into employee(id,name,age,department)
Values (101, 'Abc' , 25 , 'engineer')
''')
```
