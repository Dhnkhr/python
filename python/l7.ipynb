```python
students={
    "std1":{"name":"abc","age":20},
    "std2":{"name":"cde","age":30}
}

print(students['std2']['name'])
```

```python
for std_id,std_info in students.items():
    print(f"{std_id} : {std_info}")
    for key , value in std_info.items():
          print(f"{key} : {value}")
```

```python
numbers=[1,1,2,2,2,3,4,4,4,4,5]
freq={}

for number in numbers:
    if number in freq:
        freq[number]+=1
    else:
        freq[number]=1
print(freq)
```
