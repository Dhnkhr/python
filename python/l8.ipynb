```python
def print_numbers(*args):
    for numbers in args:
        print(numbers)
        
```

```python
print_numbers(1,2,3,4,5,6,7,8)
```

```python
print_numbers(1,2,3,4)
```

```python
print_numbers(1,2,3,"abc")
```

```python
def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key} : {value}")
```

```python
print_details(name = "abc" , age = "20")
```
