```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('customers.csv')
```

```python
df.head()
```

```python
df['Date_of_Birth']=pd.to_datetime(df['Date_of_Birth'])
```

```python
df['year']=df['Date_of_Birth'].dt.year
```

```python
df.drop(columns=['Date_of_Birth'], inplace=True)
```

```python
df['fname']=df['Customer_Name'].str.split(' ').str[0]
df['lname']=df['Customer_Name'].str.split(' ').str[-1]
```

```python
df.head()
```

```python
df.tail()
```

```python
df.rename(columns={'year': 'birth_year'}, inplace=True)
```

```python
df['reg_year']=df['Registration_Date'].str.split('-').str[0]
```

```python
df['reg_year'].astype(int)
```

```python
df['Age_Group'] = df['Age_Group'].astype(str).str.replace(r'\+', '', regex=True)
```

```python
df['min_age']=df['Age_Group'].str.split('-').str[0]
df['max_age']=df['Age_Group'].str.split('-').str[1]
```

```python
df['max_age'].isnull()
```

```python
df['max_age']=df['max_age'].fillna(100)
```

```python
df[['max_age','min_age']].astype(int)
```

```python
df.head()
```

```python
df['Customer_Tier'].unique()
```

```python
df['Customer_Tier']=df['Customer_Tier'].map({'Gold':1,'Silver':0,'Platinum':2})
```

```python
df.head()
```

```python
df['phone'] = df['Phone'].str.replace(r'^\+91-', '', regex=True)
```
